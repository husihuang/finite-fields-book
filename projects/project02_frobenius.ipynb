```python
class GF2m:
    """GF(2^m) 的教学用位编码实现。"""
    def __init__(self, modulus):
        self.modulus = modulus
        self.m = modulus.bit_length() - 1
        self.q = 1 << self.m

    def add(self, a, b):
        return a ^ b

    def mul(self, a, b):
        result = 0
        while b:
            if b & 1:
                result ^= a
            b >>= 1
            a <<= 1
            if a & self.q:
                a ^= self.modulus
        return result

    def power(self, a, n):
        result = 1
        while n:
            if n & 1:
                result = self.mul(result, a)
            a = self.mul(a, a)
            n >>= 1
        return result

F = GF2m(0b1011)  # x^3+x+1
assert F.mul(2, 4) == 3
print('GF(8) 已建立')
```
